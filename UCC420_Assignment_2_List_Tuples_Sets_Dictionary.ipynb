## 1. List

Create a list L as:
`[10, 20, 30, 40, 50, 60, 70, 80]`

### i. Add 200 and 300 to L


```python
L = [10, 20, 30, 40, 50, 60, 70, 80]

L.append(200)
L.append(300)

print(L)
```

### ii. Remove 10 and 30 from L


```python
L = [10, 20, 30, 40, 50, 60, 70, 80]

L.remove(10)
L.remove(30)

print(L)
```

### iii. Sort L in ascending order


```python
L = [10, 20, 30, 40, 50, 60, 70, 80]

L.append(200)
L.append(300)

L.sort()

print(L)
```

```python
L = [10, 20, 30, 40, 50, 60, 70, 80]

L.append(200)
L.append(300)

L.sort(reverse=True)

print(L)
```

## 2. Tuple

Create:
`scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)`

### i. Identify the highest score and its index


```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

highest = max(scores)

print("Highest score --> ", highest)
print("Index --> ", scores.index(highest))
```

### ii. Find the lowest score and count how many times it appears


```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

lowest = min(scores)

print("Lowest score --> ", lowest)
print("Count --> ", scores.count(lowest))
```

### iii. Reverse the tuple and return it as a list


```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

result = list(scores[::-1])

print(result)
```

### iv. Check if a score entered by the user is present in the tuple


```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

x = float(input("Enter score to search: "))

if x in scores:
    print("Score is present")
    print("First occurrence index --> ", scores.index(x))
else:
    print("Score is not present")
```

## 3. Random Numbers

Create a list of 100 random numbers between 100 and 900 and count and print all odd, even and prime numbers.


```python
import random

numbers = []

for i in range(100):
    numbers.append(random.randint(100, 900))

odd = []
even = []
prime = []

for n in numbers:
    if n % 2 != 0:
        odd.append(n)
    else:
        even.append(n)

    f = 0

    if n > 1:
        for j in range(2, n // 2 + 1):
            if n % j == 0:
                f = 1
                break

        if f == 0:
            prime.append(n)

print("Odd numbers =", len(odd))
print(odd)

print("Even numbers =", len(even))
print(even)

print("Prime numbers =", len(prime))
print(prime)
```

## 4. Sets

A = {34, 56, 78, 90}  
B = {78, 45, 90, 23}


### i. Find the unique scores achieved by both teams (Union)


```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

print("Union --> ", A.union(B))
```

### ii. Identify the scores common to both teams (Intersection)


```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

print("Intersection --> ", A.intersection(B))
```

### iii. Find the scores exclusive to each team (Symmetric Difference)


```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

print("Symmetric Difference --> ", A.symmetric_difference(B))
```

### iv. Check if A is a subset of B and B is a superset of A


```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

print("A is subset of B --> ", A.issubset(B))
print("B is superset of A --> ", B.issuperset(A))
```

### v. Remove a specific score X from set A if it exists
