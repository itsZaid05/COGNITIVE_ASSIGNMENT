```python
n = int(input("Enter a No: "))

s = 0

for i in range(2, n + 1):
    f = 0

    for j in range(2, i // 2 + 1):
        if i % j == 0:
            f = 1
            break

    if f == 0:
        s = s + i

print("Sum of Prime Numbers --> ", s)
```

```python
def AddOdd(n):
    s = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            s = s + i

    return s


n = int(input("Enter a No: "))

print("Sum of Odd Numbers --> ", AddOdd(n))
```

```python
def IsPrime(n):
    if n < 2:
        return 0
    
    for i in range(2, n // 2 + 1):
        if n % i == 0:
            return 0

    return 1

def AddPrime(n):
    s = 0

    for i in range(2, n + 1):
        if IsPrime(i):
            s = s + i

    return s
n = int(input("Enter a No: "))

print("Sum of Prime Numbers  ", AddPrime(n))
```
